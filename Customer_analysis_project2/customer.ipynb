# check the most customers by gender

```python
df.groupby(['Gender'])['CustomerID'].sum()
```

mall has more female customers then male

```python
plt.title('Gender distbution')
sb.countplot(x=df['Gender'])

plt.xlabel('Gender')
```

```python
gen = df['Gender'].value_counts()
plt.pie(gen , labels=['Female','Male'],autopct='%1.1f%%')
plt.title('Gender')
plt.figure(figsize=(3,3))
plt.show()
```

check the income of customers by gender

```python
df.groupby(['Gender'])['Annual Income (k$)'].mean()
```

```python
df.groupby(['Gender'])['Annual Income (k$)'].mean().plot(kind='bar',color=['orange','cyan'])
plt.title('Annual income by gender')
```

```python
# check spendings by gender
df.groupby('Gender')['Spending Score (1-100)'].mean()
```

```python
plt.title('Spending 1 - 100 by Gender')
plt.ylabel('counts')
plt.xlabel('Gender')
sb.barplot(x= df['Gender'],y=df['Spending Score (1-100)'])
```

female customers has more spendings

```python


# draw the scatter show income vs spending score scatter plot by gender
sb.scatterplot(data= df,x='Annual Income (k$)',y='Spending Score (1-100)',hue='Gender')
plt.title('Annual Income vs Spending Score')
```

##Annual Income vs Spending Score Analysis

This scatter plot reveals five distinct customer segments based on income and spending behavior:

High Income, High Spending – premium/target customers, most valuable for the business

High Income, Low Spending – potential customers who could be converted through targeted offers

Low Income, High Spending – impulsive spenders, financially riskier segment

Low Income, Low Spending – budget-conscious, cautious customers

Average Income, Average Spending – the majority of customers fall into this typical/standard segment

These insights can help the business design targeted marketing strategies for each customer group. 



```python
sb.scatterplot(data = df,x='Age', y='Spending Score (1-100)',hue='Gender')
plt.title('Age vs spending score')
```

##Age vs Spending Score Analysis##

This scatter plot examines the relationship between customer age and their spending behavior:

Younger customers (roughly ages 18–35) show a wide range of spending scores, including many high scorers, indicating this age group tends to spend more freely.

Older customers (40+) generally show more moderate to lower spending scores, with less variation.

This suggests that age is a meaningful factor in predicting spending behavior — younger customers represent a more active and potentially higher-value spending segment.

Business takeaway: Marketing campaigns targeting younger age groups are likely to yield better engagement and spending response. 



 **Business Analysis Summary**

This analysis examined 200 mall customers based on their gender, age, annual income, and spending behavior to uncover patterns that can guide business strategy.

Key Findings:

Gender: The customer base is fairly balanced between Male and Female, with Female customers showing a slightly higher average spending score.

Age: Most customers fall between their 20s and 40s, with a large concentration around the 28–32 age group. Younger customers tend to have higher spending scores compared to older ones.

Annual Income: The majority of customers earn between 48k–60k, representing the core income segment the business serves.

Income vs Spending Relationship: Customers cluster into five distinct segments — high income/high spending (premium customers), high income/low spending (untapped potential), low income/high spending (impulsive spenders), low income/low spending (budget-conscious), and average income/average spending (the largest group).

Business Recommendations:

Marketing efforts should prioritize the High Income–High Spending segment, as they are the most valuable and profitable customers.

Targeted promotions or loyalty offers could help convert the High Income–Low Spending group into active spenders.

Since younger customers show higher spending scores, campaigns aimed at the 20–35 age range are likely to be more effective.

Understanding the dominant middle-income segment can help tailor pricing and product offerings to match what the majority of customers can afford.

Overall, this analysis provides a clear, data-driven foundation for designing customer-focused marketing and sales strategies.

```python


df = pd.read_csv('dirty_cafe_sales.csv')
```

```python
# check for null
df.isna().sum()
```

```python
# clean item col
df['Item'].value_counts()
```

```python
# replace unknown to NaN
df['Item'] =df['Item'].replace('UNKNOWN',np.NaN)
```

```python
df['Item'] = df['Item'].replace('ERROR',np.NaN)
```

```python
df['Item'].isna().sum()
```

```python
# fill the null values
df['Item'] = df['Item'].fillna('Unknown')
```

```python
# clean column 2 qunatity
df['Quantity'].value_counts()
```

```python
# replace
df['Quantity'] = df['Quantity'].replace('UNKNOWN',np.NaN)
df['Quantity'] = df['Quantity'].replace('ERROR',np.NaN)
```

```python
# change data type
df['Quantity'] = pd.to_numeric(df['Quantity']  )
df['Quantity'].dtype
```

```python
df['Quantity'].median()
```

```python
# FILL NAN VALUES WITH MEDIAN
df['Quantity'] = df['Quantity'].fillna(df.groupby(['Item'])['Quantity'].transform('median'))
```
